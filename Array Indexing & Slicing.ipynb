```python
import numpy as np
```

NUMPY SLICING AND INDEXING OF VECTORS

```python
arr = np.arange(11,21)
arr
```

```python
arr_slice = arr[1:7:2]  #Slicing [start:stop+1:step]
arr_slice
```

NUMPY SLICING AND INDEXING OF MATRIX

```python
arr = np.arange(1,31).reshape((6,5))
arr
```

```python
arr[5]  #Prints the row of index 5
print(arr[0,0])   #Prints the first element [row index, column index]
```

```python
slice = arr[0:2,1:3]  #Slices matrix [row_index_start: row_index_stop+1, column_index_start, column_index_stop+1]
slice
```

```python
slice2 = arr[3:,3:]  #not giving stop means it goes till the end
slice2
```

```python
slice3 = arr[0:,0:1]  #Column Slicing
slice3
```
