Deep and Shallow Copy

```python
a = np.arange(1,21)
a
```

```python
slice = a[:5]
slice = slice * 10
slice
```

```python
a     #Shallow Copy(happens when sliced because it takes up another location.)
```

```python
b = a   #DEEP COPY
b[0] = 99
b
```

```python
a  # a got changed by changing the value in b because a was assigned in b
```

```python
c = np.arange(1,21)
c
```

```python
d = c.copy()  #Shallow Copy
d[0] = 99
d
```

```python
c
```

ADVANCE ARRAY MANIPULATION

Stacking Array

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
```

```python
np.vstack((a,b))  #VERTICAL STACK
```

```python
np.hstack((a,b))   #HORIZONTAL STACK
```

```python
np.column_stack((a,b))   #COLUMN STACK
```

Splitting Arrays

```python
c = np.arange(16).reshape(4,4)
c
```

```python
np.hsplit(c,2)   #Horizontal Split
```

```python
np.vsplit(c,4)   #Vertical Split
```
